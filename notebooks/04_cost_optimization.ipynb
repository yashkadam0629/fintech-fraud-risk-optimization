Stage 3B and 3C: replace the arbitrary 0.50 threshold with one derived from an explicit cost model, then show how far it moves when the assumptions move.

All monetary figures are **SIMULATED under stated assumptions**.


```python
import sys, warnings
sys.path.insert(0, '..')
warnings.filterwarnings('ignore')
import pandas as pd
from src import config, threshold_optimization, sensitivity_analysis, plots

# Reuse the sweep the pipeline already produced, so the notebook and the
# reports cannot disagree.
sweep = pd.read_csv('../outputs/model_results/threshold_sweep_test.csv')
sweep.head()
```

```python
optimum = threshold_optimization.optimal_threshold(sweep)
{k: v for k, v in optimum.items() if not isinstance(v, (dict, str))}
```

## The cost curve


```python
import matplotlib.pyplot as plt
fixed = sweep[sweep['grid_source'] == 'fixed']
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(fixed['threshold'], fixed['total_cost'], 'k-', lw=2, label='total simulated cost')
ax.plot(fixed['threshold'], fixed['cost_missed_fraud'], 'r--', label='missed fraud')
ax.plot(fixed['threshold'], fixed['cost_customer_friction'], 'b--', label='customer friction')
ax.axvline(optimum['optimal_threshold'], color='orange', lw=2)
ax.axvline(0.5, color='grey', ls=':')
ax.set_xlabel('intervention threshold'); ax.set_ylabel('simulated cost')
ax.legend(); plt.show()
```
